#### Create a numpy array of each column and then convert that data into Panda Dataframe

```python
#creating seperate arrays for each column

order_id = np.arange(1,100001,1)
product_id = np.random.randint(1,101,100000)
customer_id = np.random.randint(1,10001,100000)
product_price = np.random.randint(5,501,100000)
discount = np.random.randint(0,31,100000)
qty = np.random.randint(1,6,100000)


modified_discount = discount/100
revenue = qty*product_price * 1-modified_discount
```

```python
#creating data frame from numpy arrays 

df = pd.DataFrame({"order_id":order_id,"product_id":product_id,"customer_id":customer_id,"product_price":product_price,"discount":discount,"qty":qty,"revenue":revenue})
```

```python
# view the first 5 entries of dataset
df.head(5)
```

```python
#basic operations 
df.count()
a= df.product_price.mean()
int(a)
df.min()
df.max()
```

```python
# Display every order where revenue is greater than 500 Euros

revenue_greater_500 = df[df["revenue"]>500]
revenue_greater_500
```

```python
#Find all orders where revenue > €500 AND discount >= 20%.

revenue_greater_500_discount_less_20 = df[(df["revenue"]>500) & (df["discount"]<=20)]
revenue_greater_500_discount_less_20
```

```python
#Give me the 10 highest-revenue orders.

sorted_data = df.sort_values("revenue",ascending=False)
sorted_data.head(10)
```

```python
customer_revenue = df.groupby("customer_id")["revenue"].sum()
#top 10 customer reveneue
customer_revenue=pd.DataFrame(customer_revenue)
customer_revenue.sort_values("revenue",ascending=False)
customer_revenue.head(10)
```

```python
# For each customer, tell me: total revenue,average order revenue,number of orders
x = df.groupby("customer_id")["revenue"].agg(['sum','mean','count'])
x
```

```python
#For each product, how many units were sold and how much revenue did it generate?

sold_units = df.groupby("product_id")["qty"].sum()
revenue_t = sold_units*df.product_price
product_stats = df.groupby("product_id").agg({
    "qty": "sum",
    "revenue": "sum"
})
```

```python
top_10_p= product_stats.sort_values("revenue",ascending=False).head(10)
```
