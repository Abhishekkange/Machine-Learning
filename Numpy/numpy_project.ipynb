# Dataset Requirements

Create a dataset containing **1,00,000 orders** with the following columns:

| Column | Description |
|---|---|
| `order_id` | Unique ID for each order |
| `customer_id` | ID of the customer |
| `product_id` | ID of the product |
| `product_price` | Price of the product in euros |
| `qty` | Quantity ordered |
| `discount` | Discount percentage |

## Assumptions

- **Total Orders:** 1,00,000
- **Total Customers:** 10,000
- **Total Products:** 100
- **Quantity (`qty`):** 1–5
- **Product Price (`product_price`):** €5–€500
- **Discount (`discount`):** 0%–30%

```python
import numpy as np
```

```python
order_id = np.arange(1,100001,1)
product_id = np.random.randint(1,101,100000)
customer_id = np.random.randint(1,10001,100000)
product_price = np.random.randint(5,501,100000)
discount = np.random.randint(0,31,100000)
qty = np.random.randint(1,6,100000)


```

```python
print(order_id)
print("----------")
print(product_id)
print("----------")
print(customer_id)
print("----------")
print(product_price)

print("----------")
print(qty)
print("----------")

print(discount)
```

#Revenue calculation 

Revenue=Quantity×Price×(1−Discount/100)

```python
modified_discount = discount/100
revenue = qty*product_price * 1-modified_discount
```

```python
revenue
```

```python
total_revenue = revenue.sum()
print("Total Revenue :",int(total_revenue))

average_order_revenue = int(total_revenue)/100000
print("Average Revenue : ",int(average_order_revenue))
```

```python
# all transcation where revenue after discount was more than 500 euros

revenue_more_500 = revenue[revenue>500]
revenue_more_500
total_such_order = revenue_more_500.size
total_such_order
```

```python
# which order generated revenue more than 500

mask1 = revenue>500
print(order_id[mask1])
```

```python
#Find orders where revenue was greater than €500 AND the discount was less than 10%.
mask2 = (revenue>500) & (discount<10)
print(revenue[mask2])
```

```python
revenue.argmax()
```

```python
#What are the 10 highest-revenue orders, and what are their order IDs?

mask3 = revenue.argsort()
inds = mask3[-10:]
order_id[inds]
```

```python
#Which 10 customers generated the most revenue?

all_customer_revenue = np.bincount(customer_id,weights=revenue)
all_customer_revenue_sorted_map = np.argsort(all_customer_revenue)
top_10_customer =all_customer_revenue_sorted_map[-10:]
top_10_customer
```
